```python
for x in range(10):
    if x == 5:
        break
    print(f'{x}-Kamran')
```

```python
for x in range(10):
    if x == 6:
        continue
    print(f'{x}-Kamran')
```

```python
count = 0
for x in ['apple', 'banana', 'cherry', 'mango', 'apple', 'grapes']:
    if x == 'apple':
        count = count + 1
        if count == 2:
            continue
    print(f'{x}-Kamran')
```

```python
# Nested Loops
for i in range(5):
    for j in ['apple', 'banana', 'cherry']:
        print(f'{i}- {j}')
    print()
```

```python
even = []
for i in range(21):
    if i % 2 == 0:
        even.append(i)

print(even)
```

```python
# List Compreshension
x = [i for i in range(20) if i % 2 == 0]
print(x)
```
