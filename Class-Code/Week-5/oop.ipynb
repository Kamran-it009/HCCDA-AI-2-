```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"Book: {self.title}"

    def __call__(self):
        print(f"Hello, {self.author}!")

    def __repr__(self):
        return f"Info({self.title}, {self.author})"
```

```python
b1 = Book('Database system', 'Thomas Conolly')
```

```python
import time

def boil_water():
    print("Boiling water...")
    time.sleep(3)
    print("Water is ready!")

def toast_bread():
    print("Toasting bread...")
    time.sleep(2)
    print("Bread is toasted!")

def make_breakfast():
    boil_water()
    toast_bread()
    print("Breakfast is ready!")

start_time = time.time()
make_breakfast()
end_time = time.time()

print(f"Total time taken: {end_time - start_time:.2f} seconds")
```

```python
import asyncio


async def boil_water():
    print("Boiling water... (3 seconds)")
    await asyncio.sleep(3)
    print("Water is ready!")

async def toast_bread():
    print("Toasting bread...(2 seconds)")
    await asyncio.sleep(2)
    print("Bread is toasted!")

start_time = time.time()  # Start the timer
async def make_breakfast():
    # Run both tasks at the same time
    await asyncio.gather(
        boil_water(),
        toast_bread()
    )
    print("Breakfast is ready!")

start_time = time.time()  # Start the timer
await make_breakfast()
end_time = time.time()  # End the timer
print(f"Total time taken: {end_time - start_time:.2f} seconds")
```

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
# This is the function we want to decorate
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```
